# 神经网络参数的初始化方式
训练开始时神经网络中神经元的参数需要有初始值. 好的初始化方式应当控制参数的分布(方差)在训练过程中保持稳定, 避免梯度消失或梯度爆炸问题.
## 常见的初始化方式
除了均匀分布初始化, 正态分布初始化, 固定值初始化 (一般用于偏置项参数 $b$ 全部设为0) 之外, 最常用的是 **Xavier 初始化和 Kaiming 初始化**.
- **Xavier 初始化**: 适用于 Sigmoid, Tanh 等激活函数, 其初始化的方差为 $\frac{2}{n_{in}+n_{out}}$, 其中 $n_{in}$ 为输入神经元的数量, $n_{out}$ 为输出神经元的数量.
  - 代入到均匀分布: $w\sim U\left[-\sqrt{\frac{6}{n_{in}+n_{out}}}, \sqrt{\frac{6}{n_{in}+n_{out}}}\right]$
  - 代入到正态分布: $w\sim N\left(0, \sqrt{\frac{2}{n_{in}+n_{out}}}\right)$
- **Kaiming 初始化**: 适用于 ReLU 激活函数, 其初始化的方差为 $\frac{2}{n_{in}}$, 其中 $n_{in}$ 为输入神经元的数量.
  - 代入到均匀分布: $w\sim U\left[-\sqrt{\frac{6}{n_{in}}}, \sqrt{\frac{6}{n_{in}}}\right]$
  - 代入到正态分布: $w\sim N\left(0, \sqrt{\frac{2}{n_{in}}}\right)$

PyTorch 会默认根据激活函数的类型选择合适的初始化方式, 例如 ReLU 激活函数会使用 Kaiming 初始化, Sigmoid 激活函数会使用 Xavier 初始化. 也可以手动指定初始化方式.

```python
# Xavier 初始化
layer = nn.Linear(5, 3) # 输入5个激活值, 输出3个激活值

# 通过 weight 和 bias 属性访问参数
# 📌nn.Linear 默认使用 Kaiming 均匀分布初始化
print(layer.weight)
print(layer.bias)

# 偏置项参数全部设为0
nn.init.zeros_(layer.bias)
# 或者 nn.init.constant_(layer.bias, 0)
print('偏置项参数全部设为0: \n', layer.bias.detach())

# Xavier 均匀分布初始化
nn.init.xavier_uniform_(layer.weight)
print('Xavier 均匀分布初始化: \n', layer.weight.detach())
# Xavier 正态分布初始化
nn.init.xavier_normal_(layer.weight)
print('Xavier 正态分布初始化: \n', layer.weight.detach())
```

```python
# Kaiming 均匀分布初始化
nn.init.kaiming_uniform_(layer.weight)
print('Kaiming 均匀分布初始化: \n', layer.weight.detach())

# Kaiming 正态分布初始化
nn.init.kaiming_normal_(layer.weight)
print('Kaiming 正态分布初始化: \n', layer.weight.detach())
```
