## 改变数组形状

### 重塑形状（reshape）

```python
a = np.array([ [0, 1, 2, 3], [4, 5, 6, 7]]) # 创建一个 2×4 的二维数组
print ("a =", a)
b = a.reshape(4, 2) # 将形状改为 4×2
print ("b =", b)
```

### 原地调整形状或大小（resize）

```python
print("a = ", a)
a.resize(4, 2) 
print ('调整形状后，a =', a)
```

## 索引与切片

### 一维数组的索引与切片

```python
a = np.arange(0, 8, 2) # 创建一个包含 4 个偶数的一维数组
print (a)
print (a[2]) # 索引: 使用索引 2 访问第 3 个元素
print (a[:]) # 切片: 访问所有元素
print (a[1:3]) # 切片: 访问索引 1 到 2 的元素，不包含结束索引 3
print (a[::-1]) # 切片: 以相反顺序访问所有元素
```

### 二维数组的索引与切片

```python
# 多维数组的索引与切片
b = np.arange(0, 24, 2).reshape(3, 4)
print (b)
print (b[1, 2])  # 索引: 用逗号分隔各轴的索引，访问第 2 行、第 3 列的元素
print (b[0:2, 1]) # 切片: 返回第 2 列前两行的元素（行索引为 0、1）
print (b[:, 1]) # 切片: 返回第 2 列中的元素
print (b[::-1, ::-1]) # 切片: 颠倒所有行的顺序，同时颠倒每行内元素的顺序
print (b[:, ::-1]) # 切片: 保留行的顺序，颠倒每行内元素的顺序
```

### 混合使用索引与切片

```python
a = np.arange(0, 12).reshape(3, 4)
print (a)
```

```python
print(a[1, :]) # 返回 a 的第 2 行，结果为一维数组
print(a[0:2, :]) # 返回 a 的第 1、2 行，结果为二维数组
```

```python
print(a[:, 3]) # 返回 a 每行的最后一个元素，结果为一维数组
print(a[:, 0:2]) # 返回 a 每行的前两个元素，结果为二维数组
```

### 使用整数索引数组进行索引

```python
a = np.arange(9).reshape(3, 3) # [[0,1,2],[3,4,5],[6,7,8]]
print (a[[0, 1, 2], [0, 1, 0]]) # 选取 a[0, 0]、a[1, 1] 和 a[2, 0]
print (a[[0, 2], [0, 1]]) # 选取 a[0, 0] 和 a[2, 1]
print (a[[0, 0], [2, 2]]) # 将 a[0, 2] 选取两次
```

### 使用布尔数组进行索引

```python
a = np.arange(9).reshape(3, 3) # [[0,1,2],[3,4,5],[6,7,8]]
print (a)
print (a > 0) # 返回一个与 a 形状相同的布尔数组
              # 若 a 中的元素大于 0，结果数组中对应位置的值为 True
```

```python
a = np.arange(9).reshape(3, 3) # [[0,1,2],[3,4,5],[6,7,8]]
print (a[ a<5 ]) # 返回由所有小于 5 的元素组成的一维数组
print (a[ a%2 == 0 ]) # 返回由所有偶数元素组成的一维数组
```

## 广播（broadcasting）

```python
a = np.arange(1, 13).reshape(4, 3)
print (a)
b = np.array([1,0,1])
print (b)
```

```python
x = a + b # 通过广播，将向量 b 加到矩阵 a 的每一行上
print (x)
```

- 上面的广播运算与下面显式编写循环的结果相同。
- 矩阵较大时，使用广播通常比显式的 Python 循环更高效。

```python
y = np.zeros((4, 3), dtype=int)
for i in range(4):
    y[i, :] = a[i, :] + b
print (y)    
```

## 改变数组形状的更多方法
### 将数组展平为一维数组

```python
a = np.array([ [0, 1, 2, 3], [4, 5, 6, 7]] ) 
print ("a = ", a)
b = a.flatten()
print ("b =", b)
```

```python
print ("a = ", a)
c = np.ravel(a)
print ("c =", c)
```

```python
b[0] = 123
c[1] = 234
print("a=", a)
```

### 数组转置

```python
a = np.array([ [0, 1, 2, 3], [4, 5, 6, 7]]) # 2×4 数组
print ("a =", a)
print ('形状 = ', a.shape)

print (a.transpose()) # 二维数组的常规转置
# 等价写法: print(np.transpose(a))

print (a.T)
```

```python
a = np.array([ [0, 1, 2, 3], [4, 5, 6, 7]]) # 2×4 数组
print ("a =", a)
print ('形状 = ', a.shape)

b1 = a.transpose(1, 0) # 按轴顺序 [1, 0] 对二维数组进行转置
                       # 即将形状从 (2, 4) 变为 (4, 2)
print ("b1 =", b1)
```

### 翻转数组

```python
a = np.array([ [0, 1, 2, 3], [4, 5, 6, 7]]) # 2×4 数组
print ("a =", a)
print ('形状 = ', a.shape)

b = np.flip(a)
print ("b = ", b)
print ('形状 = ', b.shape)
```

### 重塑形状、转置与翻转的区别

```python
a = np.arange(12).reshape(3, 4) # 数组方法 numpy.ndarray.reshape() 与函数 numpy.reshape() 对应
print ("a =", a)
print ('形状 = ', a.shape)

b = a.reshape(4, 3)
print ("b =", b)

c = a.transpose() # 数组方法 numpy.ndarray.transpose() 与函数 numpy.transpose() 对应
print ("c =", c)

d = np.flip(a) # 使用 numpy.flip() 翻转元素顺序
print ("d = ", d)
```

### 按指定的轴顺序进行转置

```python
a = np.arange(24).reshape(2, 3, 4) # 2×3×4 数组
print ("a =", a)
print ('形状 = ', a.shape)
b1 = a.transpose(2, 1, 0) # 按轴顺序 [2, 1, 0] 对三维数组进行转置
                       # 即将形状从 (2, 3, 4) 变为 (4, 3, 2)
print ("b1 =", b1)

b2 = a.transpose(1, 2, 0) # 按轴顺序 [1, 2, 0] 对三维数组进行转置
                       # 即将形状从 (2, 3, 4) 变为 (3, 4, 2)
print ("b2 =", b2)
```

### 为数组添加新的维度

```python
a = np.arange(4)
print ('a 的形状=', a.shape)
print ("a=", a)

b = a[np.newaxis, :]
print ('b 的形状=', b.shape)
print ("b=", b)

c = a[:, np.newaxis]
print ('c 的形状=', c.shape)
print ("c=", c)
```

```python
a = np.arange(4)
print ('a 的形状=', a.shape)
print ("a=", a)

b = np.expand_dims(a, axis=0)
print ('b 的形状=', b.shape)
print ("b=", b)

c = np.expand_dims(a, axis=1)
print ('c 的形状=', c.shape)
print ("c=", c)
```

## 副本与视图

```python
a = np.arange(6).reshape(2, 3) # [[0,1,2],[3,4,5]]
print (a)

b = a # b 与 a 引用同一个数组对象，没有创建新数组
b[0, 2] = 100 # 修改 b 中的一个元素
print (a) # a 中的元素也随之改变
```

```python
a = np.arange(9).reshape(3, 3) # [[0,1,2],[3,4,5],[6,7,8]]
b = a[:2] # b 是新的数组视图，内容为 [[0,1,2],[3,4,5]]，与 a 共享底层数据
b[0, 2] = 100 # 修改 b 中的一个元素
print (a) # a 中的元素也随之改变
```

```python
a = np.arange(6).reshape(2, 3) # [[0,1,2],[3,4,5]]
b = a.copy() # b 是具有独立数据的新数组副本
print (b.base)

b[0, 2] = 100 # 修改 b 中的一个元素
print (a) # 只有 b 被修改，a 保持不变
```

```python
a = np.arange(6).reshape(2, 3) # [[0,1,2],[3,4,5]]
b = a.view() # b 是新的数组视图，与 a 共享底层数据
print (b.base)

b[0, 2] = 100 # 修改 b 中的一个元素
b.resize(3, 2) # 改变 b 的形状
print (a) # a 的数据随之改变，但形状保持不变
```

### 高级索引

```python
a = np.array([[1, 2, 3], [4, 5, 6]])

# 整数数组索引
c = a[ [0, 1], [0, 2]]
print(c.base)
c[0] = 123
print("a=", a)
print("c=", c)

# 布尔数组索引
b = a[ a > 3 ]
print(b.base)
b[0] = 123
print("a=", a)
print("b=", b)
```

## 数据类型

```python
w = np.array([0.5, 1.5, 2.5], dtype=np.double) # 创建元素类型为双精度浮点数的数组
print (w)
```

```python
a = w.astype(np.int_) # 返回转换为指定数据类型的数组副本
print (a)
```

```python
z = np.arange(5, dtype=np.int_) # 创建整数数组
print (z)
```

```python
x = np.double([1, 2, 3]) # 双精度浮点数
print (x)
print (type(x[0]))
```

```python
y = np.intc([0.5, 1.5, -2.5]) # 与 C 语言 int 类型对应的有符号整数
print (y)
print (type(y[0]))
```
